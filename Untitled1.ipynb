```python
# while loop 1
waktu = 10
while waktu > 0:
  print(waktu)
  waktu -= 1
print("Selesai!")
```

```python
# while loop 2
bilangan = 2
while bilangan <= 10:
  print(bilangan)
  bilangan += 1
```
